## Metodología
Para resolver el problema se tendrá en cuenta el algoritmo PAM (Partitioning Around Medoids) que consiste en una variación más robusta de K-Medoids y cuya métrica de distancia permite la presencia de variables diferentes a numéricas. Los pasos del algoritmo son los siguientes: 

**Paso 1.** Se escoge un centro aleatorio y se asignan todos los puntos a un grupo donde guarde la menor distancia con el centro escogido (usualmente la métrica es la distancia de Manhattan). Se calcula el costo total, que se define como la sumatoria de los costos de cada punto hacia su centroide. Este valor se interpreta como el costo que tendría en modelo para moverse de un centroide a otro.  

**Paso 2.** Se selecciona un dato aleatorio dentro de la muestra para que se comporte como un nuevo centro y se calcula el costo de moverse hacia el nuevo punto. Si la diferencia entre el costo actual y el costo anterior es mayor a cero, no valdrá la pena cambiar de centro pero de lo contrario se cambiará al nuevo centro y se pasa nuevamente en paso 1. 


**Paso 3.** El algoritmo converge cuando no se cambia de centro.

Una de las ventajas de este algoritmo es que suele entregar grupos homogeneos en datasets que no sean de gran tamaño y se utiliza ampliamente dentro de problemas que involucran variables categóricas. 

Puede encontrarse mayor información sobre el algoritmo y su implementación [aquí](https://www.cs.umb.edu/cs738/pam1.pdf)

## PAM

### Preprocesamiento
Cuando se tienen mas de dos dimensiones en un problema de agrupamiento, el reto recae sobre la visualización e interpretación de los grupos, sobretodo cuando se trata con diversas variables categóricas. En este caso, es necesario utilizar para la medida de las distancias entre puntos una estandarización de las variables categóricas, a través de la función *Daisy* que utiliza una medida denominada "gower". Dicha medida admite variables diferentes a numéricas y les asigna la mejor métrica dependiendo de su tipo, por ejemplo, para las numéricas utiliza la distancia de Manhattan y para las categóricas las asume como variables binarias calculando métricas como variables dummy, de la misma forma es posible que a los factores se les asigne un peso de acuerdo a su contribución dentro del conjunto de muestra para dicha característica. 


Puede encontrarse mayor información sobre la función *Daisy* [aquí](https://stat.ethz.ch/R-manual/R-devel/library/cluster/html/daisy.html)

```r
##
## Se cargan las librerias necesarias
##
library(dplyr)
library(Rtsne)
library(ggplot2)
library(caret)
library(lubridate)
library(class)
require(cluster)
```

### Modelo 1. Sin considerar el género del cliente

El modelo 1 no considerará los perfiles de los clientes de acuerdo a su género. A continuación se procede a preprocesar creando la matriz de distancias con la función *Daisy*:

```r
##
## se asigna la medida como "gower"
##
dist <- daisy(data, metric = "gower") 
dist_matrix <- as.matrix(dist)         #Llevamos las metricas a una matriz  
head(dist_matrix)   
```

La salida es una matriz de distancias entre cada punto y el resto de los puntos del set de datos, por lo tanto la diagonal principal debe ser cero y su dimensión de 1681x1681.

```r
length(dist_matrix)
```

En este caso, no es posible visualizar los datos debido a la alta dimensionalidad de los mismos por lo que se utiliza una técnica de reducción de dimensiones no lineales llamado **RTSNE** (T-distributed Stochastic Neighbor Embedding) el cual es ampliamente usado para la visualización de algoritmos de agrupamiento. 

Puede encontrar mayor información sobre RTSNE [aquí](https://www.rdocumentation.org/packages/Rtsne/versions/0.15/topics/Rtsne)

```r
##
## Se crea el objeto que contiene los datos
##
reduc_obj <- Rtsne(dist, is_distance = TRUE)      # se indica que la matriz contiene distancias y no registros de los datos directamente
```

```r
##
## Se guarda la información en un DataFrame que contenga las coordenadas
##
reduc_data <- reduc_obj$Y %>%
  data.frame() %>%
  setNames(c("X", "Y"))
```

En la siguiente figura no se observan aparentemenete grupos explícitos dentro de los datos cuando se realiza el redimensionamiento a 2 dimensiones.

```r
library(ggplot2)
ggplot(aes(x = X, y = Y), data = reduc_data) + geom_point(shape=21, fill="skyblue",color="dodgerblue4", size=3) +  
ggtitle("Características de los clientes") + theme(plot.title = element_text(hjust = 0.5,size=18))
```

### Entrenamiento del modelo
Se utiliza el método de la silueta para escoger el número óptimo de clusters. El análisis del coeficiente de la silueta mide qué tan bien agrupadas están las observaciones y estima una distancia promedio entre los clusters en términos de todos los grupos, es decir, que tan lejanos están los puntos de un grupo de los puntos que pertenecen a otros clusters sin tener en cuenta la distancia entre centroides. 

Un coeficiente de silueta medio se debe acercar a 1 cuando los grupos se encuentran bien agrupados, cercana a 0 cuando los puntos se encuentran entre dos o varios grupos y negativo cuando un punto se encuentra en un grupo incorrecto. En este caso, a utilizará el atributo *silinfo* y *avg.width* del objeto salida de un modelo *PAM* para estimar el hiperparámetro *k* del modelo.

Puede encontrar mayor información sobre la silueta [aquí](https://www.rdocumentation.org/packages/cluster/versions/2.0.7-1/topics/silhouette)

```r
##
## Se prueba desde 2 grupos hasta 30
##
sil <- c()
for(i in 2:30){
  pam_fit <- pam(dist,diss = TRUE,k = i)
  sil[i] <- pam_fit$silinfo$avg.width
}
```

```r
##
## Se escoge el valor máximo que en este caso corresponde a k=8
##
ggplot(data=as.data.frame(sil), aes(y=sil, x=seq(1,30,1))) +
  geom_line(linetype = "dashed")+ xlab("Número de clusters") + 
  geom_point(shape=16, fill="red",color="red", size=3) +  ggtitle("Características de los clientes") + theme(plot.title = element_text(hjust = 0.5,size=18))
```

```r
##
## MÉTODO PAM
##
pam_fit <- pam(dist, diss = TRUE, k = 8)
pam_matrix<-cbind(data,pam_fit$clustering) #Se guardan los resultados en una matriz
```
