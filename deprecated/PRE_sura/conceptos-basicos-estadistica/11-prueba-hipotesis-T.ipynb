Pruebas de Hipótesis
===

* *30 min*

Una de las aplicaciones más importantes de las funciones teóricas de probabilidad es la prueba de hipótesis. En este tipo de problemas se quiere determinar si un valor observado (estadístico) proviene de una determinada distribución de probabilidades, con el fin de aceptar o rechazar una hipótesis sobre los datos. El uso de las pruebas de hipótesis es uno de los elementos fundamentales en la construcción de modelos de aprendizaje estadístico; este punto será ejemplificado en talleres posteriores. 

Al finalizar este tutorial, usted estará en capacidad de:

* Explicar el uso de las pruebas de hipótesis.


* Explicar las similitudes y diferencias entre los valores críticos y los valores $p$.


* Aplicar pruebas de hipótesis para aceptra o rechazar hipótesis sobre los datos.


## Definición

Para definir que es una prueba de hipótesis se requiere definir primero los siguientes conceptos:

**Población:** Colección de objetos o individuos sobre los cuales se desea obtener información de interés.

**Parámetro:** Un número que resume una propiedad o característica de interés de la población entera. 

**Muestra:** Grupo obtenido aleatoriamente que es representativo de la población.

**Estadístico:** Un número que resume una propiedad o característica de la muestra. 

Ahora, una [*prueba o contrate de hipótesis*](https://es.wikipedia.org/wiki/Contraste_de_hipótesis) es un tipo de prueba estadística que tiene como fin probar una hipótesis (o supuesto), $H_0$, sobre un [*parámetro*](https://es.wikipedia.org/wiki/Parámetro_estad%C3%ADstico) o característica de la población versus una hipótesis alternativa complementaria $H_a$, a partir de una [*muestra*](https://es.wikipedia.org/wiki/Estad%C3%ADstico_muestral) de la población. 

Si $\theta$ denota un estadístico de la muestra, entonces una prueba de hipótesis es una regla que indica cuando aceptar o rechazar la hipótesis nula: 

* $H_0: \theta \sim (·)$
* $H_a: \text{ en caso contario}$ 

Por ejemplo:

* $H_0$: El paciente tiene cáncer.
* $H_a$: El paciente no tiene cancer. 

## Tipos de errores 

Se puede dar una de las siguientes situaciones:

<img src="images/pruebas-de-hipotesis/errores-tipo-I-y-II.jpg" width=200>

* Aceptar $H_0$ cuando $H_0$ es verdadera: Ok!, por ejemplo, el paciente realmente tiene cáncer y la prueba indica que tiene cáncer (se acepta la hipótesis nula); se obtiene un verdadero positivo (VP).


* Rechazar $H_0$ cuando $H_0$ es falsa: Ok!. El paciente no tiene cáncer y la prueba indica que no tiene cancer. Se obtiene un verdadero negativo (TN).


* Rechazar $H_0$ cuando $H_0$ es verdadera: Error Tipo I. El paciente tiene cáncer y la prueba indica que no tiene (Falso negativo, FN).


* Aceptar $H_0$ cuando $H_0$ es falsa: Error Tipo II. El paciente no tiene cáncer y la prueba indica que si lo padece (Falso positivo, FP)

Estos cuatro casos están estrechamente relacionados con las matrices de confusión en clasificadores, tal como se muestra a continuación:


                 |    Real
                 |  Ho    Ha
    -------------|------------  
              P  |  TP    FP          
    Decisión     |
              N  |  FN    TN
    
    TP - Verdadero positivo (correcto)
    TN - Verdadero negativo (correcto)
    FN - Falso negativo (mal clasificado)
    FP - Falso positivo (mal clasificado)

## Tipos de pruebas de hipótesis

Se consideran los siguientes tipos, donde $a$ es una cantidad numérica estimada a partir de la muestra de datos disponible:

* $H_0: \theta = a, \quad H_a: \theta < a$, cola a la izquierda.


* $H_0: \theta = a, \quad H_a: \theta \ne a$, dos colas.


* $H_0: \theta = a, \quad H_a: \theta > a$, cola a la derecha.

En la práctica, cada prueba indica a cual de los tres tipos anteriores pertenece.

En los tres casos, se desea saber si el valor $a$ está en la región blanca o en la región sombreada de las siguientes figuras. En el caso de la cola a la izquierda, se espera que si $a$ realmente no pertenece a la distribución de probabilidades estipulada, se encuentre ubicado en la región gris y por tanto se pueda rechazar la hipotesis nula; si se supone que $a$ = 1.75, entonces como está ubicado en la región blanca se acepta.

## Prueba usando valores críticos

En esta aproximación se computa el valor $z$ asociado a un nivel de confianza dado (valor crítico) que es el valor límite a partir del cual se rechaza la hipótesis nula. Por ejemplo, si la prueba es de dos colas (figura central de la parte de arriba), los valores de $z$ asociados a un nivel de confianza del 5% son $-1.96$ y $+1.96$, esto es, el área bajo la curva de la distribución normal estándar para $-1.96 \le z \le +1.96$ es 0.95, mientras que el área total de las colas sobreadas es del 0.05 (5%). Ahora, si se cumple que $-1.96 \le a \le +1.96$, entonces se acepta la hipótesis nula, y en caso contrario se rechaza. 

La prueba se interpreta de la siguiente manera: solo en el 5% de los casos no se cumplirá que $-1.96 \le a \le +1.96$ aunque la hipótesis nula sea verdadera; por ejemplo, si $a$ es un estadístico que permite determinar si una persona tiene cáncer (hipótesis nula), se espera que en 100 pacientes que tienen cáncer, se diagnostique erróneamente el 5%, que corresponde al error Tipo I. 

El procedimiento es el siguiente:


* Se definen las hipótesis nula y alternativa.


* Se computa el estadístico a partir de la muestra de datos (se calcula el valor de $a$).


* Se determina el nivel de confianza de la prueba, usualmente 10%, 5% o 1%, que corresponde a la región sombreada de las figuras.


* Se calculal el o los límites de $z$ asociados al nivel de confianza (o valor crítico definido), dependiendo si la prueba es de una cola o dos colas.

* Si el valor del estadístico $a$ cae en una de las regiones grices, se rechaza $H_0$; en caso contrario se acepta $H_0$ como verdadera.

---
**Ejemplo.--** El estadístico $a$ computado a partir de la muestra de datos es 1.75. Determine si la hipótesis nula se acepta o se rechaza para un nivel de confianza de 5% cuando:

* $\theta \sim \text{N}(0,1)$ con cola a la izquierda.
* $\theta \sim \text{N}(0,1)$ con dos colas.
* $\theta \sim \text{N}(0,1)$ con cola a la derecha.

*Solución*. Para ilustrar el proceso de solución se usará la figura de abajo. 


* Para la figura de la cola a la izquierda en el gráfico anterior se obtiene el valor de $z$ tal que el área sombreada bajo la curva sea de 0.05; esto es, $z=-1.6449$. Esto es lo mismo que decir que el estadístico de interés caerá el 95% de las veces en el área blanca. Como $a=1.75$ está en la región no sombreada, se acepta $H_0$.

```r
## area entre -inf y z tal que el area sea 5%
qnorm(0.05)
```

* Para el caso de dos colas, el área sombreada total debe ser igual a 0.05, es decir, cada una de las dos regiones sombreadas debe tener un área de 0.025. En este caso, $z=\pm 1.9600$. Ya que  $-1.96 \le a \le 1.96$ ocurre el 95% de las veces, se acepta $H_0$ para $\theta = 1.75$. 

```r
## area entre -inf y z tal que el area sea 5%/2
## los puntos +/- z son simétricos respecto al origen
qnorm(0.05/2)
```

* Para el caso de la cola a la derecha, el área sombreada debe ser igual a 0.05; esto ocurre cuando $z = 1.6449$. Es decir, el 95% de las veces ocurrirá que $-\infty \le a \le 1.6449$, entonces se rechaza $H_0$.

```r
## area entre -inf y z tal que el area blanca sea 95%
qnorm(0.95)
```

## Prueba usando el valor-$p$

Esta prueba es muy similar a la de valores críticos. En este caso se parte del estadístico $a$ y se calcula su nivel de confianza asociado tendiendo en cuenta si la prueba es de una o dos colas; en este caso, el nivel de confianza se conoce como valor-$p$. Si el valor-$p$ es menor que el nivel de confianza definido para la prueba, la hipótesis nula se rechaza, y en caso contrario se acepta. 

En esta aproximación:

* Se definen las hipótesis nula y alternativa.


* Se computa el estadístico $a$ usando la muestra de datos.


* Se calcula el valor-$p$ usando la distribución conocida que sigue el estadístico. Este es el valor del área de la distribución de probabilidades en que se rechaza la hipótesis nula.


* Se establece el nivel de significancia $\alpha$ (típicamente 0.01, 0.05 o 0.10) y se compara con el valor-$\alpha$.


* Si el valor-$p$ es menor o igual a $\alpha$ se rechaza $H_0$; en caso contario se acepta.

---
**Ejemplo.--** El estadístico $\theta$ computado a partir de la muestra de datos es 1.75. Determine si la hipótesis nula se acepta o se rechaza para un nivel de confianza de 0.05 cuando:

* $\theta \sim \text{N}(0,1)$ con cola a la izquierda.
* $\theta \sim \text{N}(0,1)$ con dos colas.
* $\theta \sim \text{N}(0,1)$ con cola a la derecha.

*Solución*. Para ilustrar el proceso de solución se usará la figura de abajo. 


* Para la cola a la izquierda se obtiene el valor-$p$ para $\theta=1.75$ (esto es, el área sombreada bajo la curva entre $-\infty$ y 1.75), o valor-$p$ = 0.9599. Ya que el valor-$p$ = 0.9599 > 0.05 (nivel de confianza) se acepta $H_0$. Véase la figura de abajo.

```r
## área entre -inf y 1.75
pnorm(1.75)
```

* Para el caso de dos colas, se computa el área sombreada bajo la curva, que corresponde a 0.0801 (valor-$p$); ya que 0.0801 > 0.05 (nivel de confianza), entonces se acepta $H_0$. Véase la figura de abajo.

```r
## área entre -inf y -1.75 mas 1.75 e +inf, simétricas
2 * pnorm(-1.75)
```

* Para el caso de la cola a la derecha, el área sombreada es 0.0401; ya que valor-$p$ = 0.0401 < 0.05 se rechaza $H_0$.

```r
## área entre 1.75 e +inf = 
## 1.0 - area entre -inf y 1.75
1.0 - pnorm(1.75)
```
