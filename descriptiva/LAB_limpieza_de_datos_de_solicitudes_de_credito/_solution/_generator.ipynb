```python
import random

import pandas as pd
import os

df = pd.read_csv("_solicitudes_de_credito.csv", sep=";")
df = df.drop_duplicates()
df = df.dropna()

df_sample1 = df.sample(frac=0.05)
df_sample2 = df.sample(frac=0.01)
df_sample3 = df.sample(frac=0.01)
df_sample4 = df.sample(frac=0.01)
df_sample2["tipo_de_emprendimiento"] = pd.NA
df_sample3["barrio"] = pd.NA
df_sample4["linea_credito"] = pd.NA

df = pd.concat([df, df_sample1, df_sample2, df_sample3])
df = df.sample(frac=1).reset_index(drop=True)

for col in [
    "sexo",
    "tipo_de_emprendimiento",
    "idea_negocio",
    "barrio",
    "línea_credito",
]:
    sample = random.sample(range(0, 10206), random.randint(100, 500))
    df.loc[sample, col] = df.loc[sample, col].str.upper()

    sample = random.sample(range(0, 10206), random.randint(100, 500))
    df.loc[sample, col] = df.loc[sample, col].str.title()

    sample = random.sample(range(0, 10206), random.randint(100, 500))
    df.loc[sample, col] = df.loc[sample, col].str.replace(" ", "-")

    sample = random.sample(range(0, 10206), random.randint(100, 500))
    df.loc[sample, col] = df.loc[sample, col].str.replace(" ", "_")


sample = random.sample(range(0, 10206), random.randint(300, 580))
df["estrato"] = df["estrato"].astype(str)
df.loc[sample, "estrato"] = df.loc[sample, "estrato"].map(lambda x: "0" + str(x))


sample = random.sample(range(0, 10206), random.randint(500, 1000))
df.loc[sample, "fecha_de_beneficio"] = df.loc[sample, "fecha_de_beneficio"].map(
    lambda x: x[-4:] + "/" + x[-7:-5] + "/" + x[-len(x) : -8]
)

sample = random.sample(range(0, 10206), random.randint(300, 600))
df["monto_del_credito"] = df["monto_del_credito"].astype(str)
df.loc[sample, "monto_del_credito"] = df.loc[sample, "monto_del_credito"].map(
    lambda x: "$ {:0,.2f}".format(float(x))
)

os.makedirs("input/", exist_ok=True)

df.to_csv("input/solicitudes_de_credito.csv", sep=";")
```
