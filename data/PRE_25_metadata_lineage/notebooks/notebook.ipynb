```python
# Problema: Documentar el origen, las transformaciones y el destino de una métrica analítica.

from pathlib import Path

import pandas as pd

ROOT = next(
    p
    for p in [Path.cwd().resolve(), *Path.cwd().resolve().parents]
    if (p / "submission").is_dir()
)
```

```python
catalog = pd.DataFrame(
    [
        ["raw_transactions", "Raw transactions", "ingestion_team", "internal"],
        ["sales_curated", "Validated sales", "analytics_team", "internal"],
        ["sales_dashboard", "Category sales", "analytics_team", "public"],
    ],
    columns=["dataset", "description", "owner", "classification"],
)
catalog
```

```python
columns = pd.DataFrame(
    [
        ["raw_transactions", "customer_id", "internal"],
        ["sales_curated", "sales_amount", "internal"],
        ["sales_dashboard", "sales_amount", "public"],
    ],
    columns=["dataset", "column_name", "classification"],
)
columns
```

```python
lineage = pd.DataFrame(
    [
        ["raw_transactions", "sales_curated", "transform and validate"],
        ["sales_curated", "sales_dashboard", "aggregate by category"],
    ],
    columns=["source_dataset", "target_dataset", "transformation"],
)
lineage
```

```python
catalog.to_csv(ROOT / "submission/data_catalog.csv", index=False)
columns.to_csv(ROOT / "submission/column_catalog.csv", index=False)
lineage.to_csv(ROOT / "submission/lineage.csv", index=False)
assert set(lineage.source_dataset) == {"raw_transactions", "sales_curated"}
```
