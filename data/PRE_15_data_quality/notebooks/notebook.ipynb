```python
# Problema: Identificar y cuantificar defectos de calidad antes de usar los datos en un análisis.

"""Evalúa cinco defectos de calidad deliberadamente controlados."""
from pathlib import Path
import pandas as pd
ROOT=next(path for path in [Path.cwd().resolve(), *Path.cwd().resolve().parents] if (path / "data").is_dir() and (path / "submission").is_dir()); DATA,OUTPUT=ROOT/"data",ROOT/"submission/quality_report.csv"
def build_submission():
    rows=[["T1","2026-04-01","C1","P1",1],["T2","2026-04-01",None,"P1",1],["T3","2026-04-01","C2","P1",0],["T4","2026-04-01","C3","PX",1],["T1","2026-04-01","C1","P1",1],["T5","2026-03-01","C4","P1",1]]
    tx=pd.DataFrame(rows,columns=["transaction_id","transaction_date","customer_id","product_id","quantity"]); products=pd.DataFrame([["P1"]],columns=["product_id"])
    tx.to_parquet(DATA/"transactions.parquet",index=False); products.to_parquet(DATA/"products.parquet",index=False)
    counts=[tx.customer_id.isna().sum(),(tx.quantity<=0).sum(),tx.transaction_id.duplicated(keep=False).sum()-1,(~tx.product_id.isin(products.product_id)).sum(),(~tx.transaction_date.eq("2026-04-01")).sum()]
    pd.DataFrame(zip(["customer_id_present","quantity_positive","transaction_id_unique","product_known","batch_fresh"],["completeness","validity","uniqueness","consistency","freshness"],counts),columns=["rule_name","dimension","violations"]).assign(status=lambda x:x.violations.map(lambda n:"FAIL" if n else "PASS")).to_csv(OUTPUT,index=False)
if __name__=="__main__": build_submission()
```
