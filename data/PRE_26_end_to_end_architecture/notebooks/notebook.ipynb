```python
# Problema: Conectar fuentes, transformaciones, calidad y consumo en una arquitectura de datos orientada a Analytics.

from pathlib import Path

import pandas as pd

ROOT = next(
    p
    for p in [Path.cwd().resolve(), *Path.cwd().resolve().parents]
    if (p / "submission").is_dir()
)
```

```python
components = pd.DataFrame(
    [
        ["transactions_db", "operational source", "sales transactions"],
        ["raw_storage", "raw layer", "preserve source copies"],
        ["curated_sales", "curated layer", "validated analytics dataset"],
        ["sales_mart", "serving layer", "management dashboard"],
    ],
    columns=["component", "layer", "purpose"],
)
components
```

```python
flows = pd.DataFrame(
    [
        ["transactions_db", "raw_storage", "batch", "preserve operational data"],
        ["raw_storage", "curated_sales", "batch", "validate and transform"],
        ["curated_sales", "sales_mart", "batch", "aggregate for consumer"],
    ],
    columns=["source", "target", "mode", "purpose"],
)
flows
```

```python
decisions = pd.DataFrame(
    [
        ["quality gate", "defective records do not reach curated", "data trust"],
        ["serving grain", "dashboard uses category aggregation", "consumer need"],
        [
            "batch processing",
            "daily sales need no real-time latency",
            "latency requirement",
        ],
    ],
    columns=["decision", "rationale", "requirement"],
)
coverage = pd.DataFrame(
    [
        ["reliable analytics data", "quality gate", "PASS"],
        ["management dashboard", "sales_mart", "PASS"],
        ["traceability", "raw_storage", "PASS"],
    ],
    columns=["requirement", "component", "status"],
)
```

```python
components.to_csv(ROOT / "submission/architecture_components.csv", index=False)
flows.to_csv(ROOT / "submission/data_flows.csv", index=False)
decisions.to_csv(ROOT / "submission/architecture_decisions.csv", index=False)
coverage.to_csv(ROOT / "submission/requirement_coverage.csv", index=False)
assert coverage.status.eq("PASS").all()
```
