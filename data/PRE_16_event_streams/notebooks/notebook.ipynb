## Solución
La celda siguiente materializa los datos y artefactos definidos por el PRE.

```python
"""Consume un log pequeño preservando llegada, estado y evento tardío."""
import csv
from pathlib import Path
ROOT=next(path for path in [Path.cwd().resolve(), *Path.cwd().resolve().parents] if (path / "data").is_dir() and (path / "submission").is_dir()); EVENTS=ROOT/"data/events.csv"; CONSUMED=ROOT/"submission/consumed_events.csv"; SUMMARY=ROOT/"submission/stream_summary.csv"
def build_submission():
    rows=[("E1","2026-01-01T10:00:00","P1",1,10),("E2","2026-01-01T10:10:00","P2",2,20),("E3","2026-01-01T10:05:00","P1",1,10),("E4","2026-01-01T10:15:00","P3",3,30)]
    with EVENTS.open("w",newline="") as f:w=csv.writer(f);w.writerow(["event_id","event_time","product_id","quantity","sales_amount"]);w.writerows(rows)
    maximum=""; consumed=[]
    for position,row in enumerate(rows,1):
        delayed=row[1]<maximum; maximum=max(maximum,row[1]);consumed.append((position,*row,delayed))
    with CONSUMED.open("w",newline="") as f:w=csv.writer(f);w.writerow(["arrival_position","event_id","event_time","product_id","quantity","sales_amount","is_delayed"]);w.writerows(consumed)
    metrics=[("event_count",len(rows)),("unique_products",len({r[2] for r in rows})),("total_sales",sum(r[4] for r in rows)),("delayed_event_count",sum(r[-1] for r in consumed))]
    with SUMMARY.open("w",newline="") as f:w=csv.writer(f);w.writerow(["metric","value"]);w.writerows(metrics)
if __name__=="__main__":build_submission()
```
