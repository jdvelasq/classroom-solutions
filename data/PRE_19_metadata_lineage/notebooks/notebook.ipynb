# PRE 19 — Saber qué datos existen y de dónde vienen

## Objetivo
Construir un catálogo mínimo y seguir el impacto de una transformación.

## Solución
Cada activo declara propósito, responsable y clasificación; el linaje explica qué depende de qué.

```python
from pathlib import Path
import pandas as pd
ROOT = next(path for path in [Path.cwd().resolve(), *Path.cwd().resolve().parents] if (path / "submission").is_dir())
catalog = pd.DataFrame([["raw_transactions","Raw transactions","ingestion_team","internal"],["sales_curated","Validated sales","analytics_team","internal"],["sales_dashboard","Category sales","analytics_team","public"]], columns=["dataset","description","owner","classification"])
columns = pd.DataFrame([["raw_transactions","customer_id","internal"],["sales_curated","sales_amount","internal"],["sales_dashboard","sales_amount","public"]], columns=["dataset","column_name","classification"])
lineage = pd.DataFrame([["raw_transactions","sales_curated","transform and validate"],["sales_curated","sales_dashboard","aggregate by category"]], columns=["source_dataset","target_dataset","transformation"])
catalog.to_csv(ROOT / "submission/data_catalog.csv", index=False)
columns.to_csv(ROOT / "submission/column_catalog.csv", index=False)
lineage.to_csv(ROOT / "submission/lineage.csv", index=False)
# A change in raw_transactions reaches both downstream assets.
assert set(lineage.source_dataset) == {"raw_transactions", "sales_curated"}
```

## Verificación
Un cambio en el dato crudo permite identificar los activos afectados.
