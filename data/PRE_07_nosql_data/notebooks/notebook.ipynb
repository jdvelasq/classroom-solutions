```python
# Problema: Comparar modelos de datos no relacionales para consultas analíticas concretas de un catálogo.

"""Genera la comparación explícita de modelos para el catálogo."""
import csv
import json
from pathlib import Path

ROOT = next(path for path in [Path.cwd().resolve(), *Path.cwd().resolve().parents] if (path / "data").is_dir() and (path / "submission").is_dir())
SOURCE, OUTPUT = ROOT / "data/products.json", ROOT / "submission/model_comparison.csv"
CRITERIA = [
    ("fixed_schema", "Necesita columnas para todos los atributos", "Conserva campos comunes y atributos anidados", "Las categorías variables no fuerzan columnas vacías"),
    ("nested_data", "Requiere separar o aplanar atributos", "Representa attributes y tags directamente", "La forma del producto permanece legible"),
    ("variable_attributes", "Genera muchas columnas irrelevantes", "Cada categoría declara solo sus atributos", "Una categoría nueva no altera documentos anteriores"),
    ("relationships", "Expresa relaciones con claves y joins", "Puede duplicar información para una lectura", "Las relaciones complejas siguen requiriendo decisión explícita"),
    ("duplication", "Reduce duplicación mediante tablas relacionadas", "Puede duplicar datos para consumo", "Se evalúa el costo de actualizar copias"),
    ("updates", "Actualiza atributos en columnas conocidas", "Actualiza el documento y su forma", "Los cambios deben preservar consistencia"),
    ("analytical_integration", "Facilita agregaciones tabulares homogéneas", "Puede requerir aplanar atributos variables", "Analytics elige la representación según la pregunta"),
]


def build_submission():
    products = json.loads(SOURCE.read_text(encoding="utf-8"))
    assert all({"product_id", "product_name", "category", "price", "attributes", "tags"} <= set(p) for p in products)
    with OUTPUT.open("w", newline="", encoding="utf-8") as file:
        writer = csv.writer(file); writer.writerow(["criterion", "relational_model", "document_model", "case_implication"]); writer.writerows(CRITERIA)


if __name__ == "__main__": build_submission()
```
