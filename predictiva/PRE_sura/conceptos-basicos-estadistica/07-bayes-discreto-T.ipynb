Noción de Probabilidad y Teorema de Bayes
===

* *60 min*

Los [clasificadores bayesianos ingenuos](https://es.wikipedia.org/wiki/Clasificador_bayesiano_ingenuo) son un tipo de clasificador probabilistico en el que se considera que cada característica de un ejemplo o instancia contribuye independientemente de las demás a que un objeto pertenezca a una clase determinada. Mientras que en la inducción de reglas de asociación (algoritmo 1R) solamente se considera una sola característica (la variable que usa el clasificador) para determinar a que clase pertence un ejemplo, en un clasificador ingenuo se consideran simultáneamente todas las características o variables de entrada. En este tutorial se describen los fundamentos matemáticos en que se soporta este tipo de clasificadores y como se aplican a casos reales. 

Al finalizar este tutorial, usted estará en capacidad de:

* Definir los conceptos de probabilidad condicional y teorema de Bayes.


* Computar probabilidades condicionales.


* Calcular la probabilidad de pertenencia de una instancia a una clase.


* Decidir a que clase pertenece una instancia a partir de su probabilidad de pertenencia a cada clase.

## Definición del problema

En este tutorial se aborda el problema de determinar si un mensaje de texto es legítimo o es spam. Este es un problema típico de minería de texto. La recepción de publicidad no deseada y mensajes fraudulentos es un problema que afecta a muchos usuarios y a las compañías que prestan el servicio de telefonía móvil; y es por ello, que las compañias prestadoras de servicios desean filtrar este tipo de mensajes con el fin de evitar el consumo de espacio en su infraestructura y la molestia para el usuario. Especificamente, se tiene una muestra conformada por los siguientes mensajes de texto, donde las variables $w_i$ representan las palabras que conforman el mensaje:


     #  Tipo      Mensaje
    ---------------------------------------
     1  spam      w1 w3 
     2  spam      w1 w2 w1 w3
     3  legítimo  w2 w4 
     4  legítimo  w4 w5 w2
     5  legítimo  w2 w4 w2


El problema consiste en clasificar si un mensaje SMS es legítimo  o spam, a partir del análisis de las palabras que contiente; se supone que ciertas palabras que son más frecuentes dependiendo del tipo de mensaje.

## Solución

### Definiciones básicas

Si las variables $x_i$ representan los eventos posibles en el universo, entonces:

* Todas las probabilidades deben estar entre $0$ y $1$: 


$$0 \le \text{Pr}(x_i) \le 1$$


* Las probabilidades de eventos mutuamente exclusivos (no pueden ocurrir simultáneamente) y colectivamente exhaustivos (cubren todo el universo de casos posibles) deben sumar la unidad:

$$\sum_{i=1}^n \text{Pr}(x_i) = 1$$

En las siguientes figuras, los eventos F1, F2 y F3, y V1 y V2 están definidos sobre el mismo universo; y son mutuamente exclusivos y colectivamente exhaustivos, tal que se cumplen las dos propiedades anteriores

<img src="images/bayes/eventos-conjuntos-2.jpg" width=500>

$$\text{Pr}(F1) + \text{Pr}(F2) + \text{Pr}(F3) = 1, \quad \qquad \text{Pr}(V1) + \text{Pr}(V2) = 1$$ 

### Probabilidad conjunta

Los eventos considerados ocurren simultáneamente. En la siguiente figura, los eventos F1 y V2 ocurren simultáneamente (área sombreada de la figura), tal que su probabilidad conjunta es:

$$\text{Pr}(F1~\text{and}~V2)$$

<img src="images/bayes/probabilidad-conjunta-3.jpg" width=250>

### Probabilidad condicional

Es la probabilidad de que ocurra un evento sabiendo que el otro ya ocurrio. En la siguiente figura el evento V2 es condicionar a la ocurrencia F1. Noté que en la siguiente figura, el universo no es el rectángulo anterior que cubre todos los eventos, si no el evento F1. De esta forma, la probabilidad condicional es sólo la proporción de V2 que se intercepta con F1, la cual corresponde a la porción sombreada de la figura de abajo. La siguiente expresión matemática permite calcular la probabilidad condicional en términos de la probabilidad conjunta.

$$\text{Pr}(V2 \; | \; F1) = \text{Pr}(F1 \; \text{and} \; V2) \; / \; \text{Pr}(F1)$$

En otras palabras, 

$$\text{Pr}(A \; | \; B) * \text{Pr}(B) = \text{Pr}(A \; \text{and} \; B)$$

para dos eventos A y B.

<img src="images/bayes/probabilidad-condicional.jpg" width=150>

### Independencia

Si los eventos $A$ y $B$ son independientes, la probabilidad condicional del evento A dado que ocurrio el evento B es igual a la probabilidad del evento A:

$$\text{Pr}(A \; |  \; B) = \text{Pr}(A)$$

De la definición de probabilidad condicional:

$$\text{Pr}(A  \; |  \; B) = \text{Pr}(A) = \frac{\text{Pr}(A\text{ and }B)}{\text{Pr}(B)}$$

Entonces:

$$\text{Pr}(A\text{ and } B) \; =  \; \text{Pr}(A) \; * \; \text{Pr}(B)$$ 

### Probabilidad marginal 

Sea $X_1$ con $i = 1, ... , n$ , un conjunto de eventos mutuamente exclusivos y colectivamente exhaustivos. La probabilidad de un evento $A$ es:

$$\text{Pr}(A) = \sum_{i=1}^n \text{Pr}(A\text{ and }X_i)$$

En la siguiente figura se puede observar que para cualquiera de los tres eventos $F_j$ (para $j=1,2,3$)

$$\text{Pr}(F_j) = \text{Pr}(F_j\text{ and }V_1) + \text{Pr}(F_j\text{ and }V_2)$$

y que para los dos eventos $V_i$ ($i=1,2$):

$$\text{Pr}(V_i) = \text{Pr}(V_i\text{ and }F_1) + \text{Pr}(V_i\text{ and }F_2) + \text{Pr}(V_i\text{ and }F_3)$$

<img src="images/bayes/eventos-conjuntos.jpg" width=750>

### Unión (OR)

Para dos eventos $A$ y $B$:

$$\text{Pr}(A \text{ or } B) = \text{Pr}(A) + \text{Pr}(B) - \text{Pr}(A\text{ and }B)$$

En la figura de abajo se observa que al unir las regiones de los eventos F1 y V2, las áreas se traslapan y por tanto la intersección se cuenta dos veces.

<img src="images/bayes/probabilidad-conjunta-3.jpg" width=250>

### Complemento o negación

$$\text{Pr}(\text{not } A) = 1 - \text{Pr}(A)$$

### Probabilidad total

La probabilidad total indica que la probabilidad de un evento A puede calcularse como la probabilidad de que ocurran los eventos A y B simultáneamente más la probabilidad de que ocurran los evento A y *not* B (el complemento de B:

$$\text{Pr}(A) = \text{Pr}(A\text{ and } B) + \text{Pr}(A\text{ and } \text{not }B) $$ 

La ecuación anterior puede expresarse en términos de probabilidades condicionales, tal que:

$$\text{Pr}(A) = \text{Pr}(A \; | \; B) \; \text{Pr}(B)  \; + \;  \text{Pr}(A \; | \; \text{not }B)\text{ Pr}(\text{not }B)$$

### Teorema de Bayes

A partir de 

$$\text{Pr}(A  \; |  \; B) = \frac{\text{Pr}(A\text{ and }B)}{\text{Pr}(B)},  
\quad 
\text{Pr}(B  \; |  \; A) = \frac{\text{Pr}(A\text{ and }B)}{\text{Pr}(A)} $$

Se obtiene que:

$$\text{Pr}(A \; | \; B)\text{ Pr}({B}) =  \text{Pr}(B \; | \; A) \; \text{Pr}({A})$$

Despejando $\text{Pr}(B \; | \; A)$,

$$\text{Pr}(B \; | \; A) = 
    \frac{\text{Pr}(A \; | \; B) \; \text{Pr}(B)}{\text{Pr}(A)} = 
    \frac{\text{Pr}(A \; | \; B)~\text{Pr}(B)} {\text{Pr}(A \; | \; B) \; \text{Pr}(B)  \; + \;  \text{Pr}(A \; | \; \text{not }B) \; \text{Pr}(\text{not }B)}$$
    
En la última ecuación, se aplica el teorema de probabilidad total para el evento A.

### Algoritmo Naive Bayes

En términos del problema de filtrado de mensajes de texto, V1 se interpreta como "Es spam" y V2 como NOT "Es spam", ya que son eventos mutuamente exclusivos y colectivamente exhaustivos. Si Fi es la ocurrencia de una determinada palabra en el texto, como por ejemplo "Viagra", entonces F1 sería "Viagra"("viagra" aparece en el mensaje) y su complemento NOT F2 sería NOT "Viagra" ("viagra" no aparece en el mensaje).   

De acuerdo con el teorema de Bayes:

$$\text{Pr}(\text{spam} \; | \; \text{viagra}) = \frac{\text{Pr}(\text{viagra} \, | \, 
\text{spam})*\text{Pr}(\text{spam})}{\text{Pr}(\text{viagra})}$$

* $\text{Pr}(\text{spam} \, | \, \text{viagra})$ es la probabilidad posterior.


* $\text{Pr}(\text{viagra} \, | \, \text{spam})$ es la verosimilitud.


* $\text{Pr}(\text{spam})$ es la probabilidad prior, es decir, la probabilidad de que un mensaje sea spam sin conocer el texto que contiene.


* $\text{Pr}(\text{viagra})$ es la verosimilitud marginal.


### Aplicación al problema de filtado de mensajes

Se tiene una muestra de ejemplos de mensajes que han sido catalogados como spam y legítimos (no spam): 

     #  Tipo      Mensaje
    -------------------------------
     1  spam      w1 w3 
     2  spam      w1 w2 w1 w3
     3  legítimo  w2 w4 
     4  legítimo  w4 w5 w2
     5  legítimo  w2 w4 w2
     

#### Matriz de términos de los documentos

Esta tabla es interpretada más fácilmente como una tabla de frecuencias que indica la cantidad de veces que una palabra aparece en un mensaje:

     #  Tipo     w1 w2 w3 w4 w5
    ----------------------------
     1  spam      1  0  1  0  0
     2  spam      2  1  1  0  0
     3  legítimo  0  1  0  1  0 
     4  legítimo  0  1  0  1  1
     5  legítimo  0  2  0  1  0
    -----------------------------
     Suma =       3  5  2  3  1
     ----------------------------
     Total de
     palabras = 3 + 5 + 2 + 3 + 1 = 14

La tabla anterior se conoce como matriz de términos del documento. Las columnas corresponden a las palabras y as filas son cada uno de los documentos. El contenido de la matriz representa la cantidad de veces que aparece cada término (palabra) en cada documento.


Para realizar la clasificación se tienen cinco palabras $w_1$, $w_2$, $w_3$, $w_4$ y $w_5$ que pueden estar o no en cada uno de los mensajes de texto. La probabilidad de que la palabra $w_1$ este en el mensaje se nota como $\text{Pr}(w_1)$, y de que no este como $\text{Pr}(\text{not }w_1)$.

Para solucionar el problema se deben cómputar las tablas descritas a continuación.

#### Probabilidad individual

Se computa la probabilidad de ocurrencia de cada palabra $w_i$ en la totatidad de palabras o $\text{Pr}(w_i)$, a partir de la tabla anterior. Para computar la probabilidad, el denominador es la suma de la cantidad de palabras por mensaje (14), y el numerador es la cantidad de veces que aparece (o no aparece) $w_i$ en el total de mensajes. 

                     w1    w2    w3    w4    w5
      ----------------------------------------------
      Ocurre       3/14  5/14  2/14  3/14   1/14
      No ocurre   11/14  9/14 12/14 11/14  13/14

#### Probabilidad conjunta $\text{Pr}(w_i, \text{Tipo})$:

Es la cantidad de veces que aparece cada palabra dependiendo del tipo de mensaje.

      Tipo        w1    w2    w3    w4    w5  Pr(tipo)
      -------------------------------------------------     
      spam      3/14  1/14  2/14  0/14  0/14      6/14 
      legítimo  0/14  4/14  0/14  3/14  1/14      8/14
     
Note que la probabilidad de que sea spam no es 2/5 (2 mensajes spam y 3 mensajes legítimos), ya que el universo no son los mensajes, si no las palabras (y su frecuencia).

#### Probabilidad condicional $\text{Pr}(w_i \, | \, \text{Tipo})$:

Es la probabilidad de que la palabra ocurra si ya se sabe que el mensaje es legítimo o es spam.

      Tipo       w1    w2    w3    w4    w5 
      --------------------------------------     
      spam      3/6   1/6   2/6   0/6   0/6   
      legítimo  0/8   4/8   0/8   3/8   1/8  

#### Probabilidad condicional $\text{Pr}(\text{not } w_i \, |  \, \text{Tipo})$:

Es la probabilidad de que la palabra NO ocurra si ya se sabe que el mensaje es legítimo o es spam. Se calcula como el complemento de la tabla anterior.

      Tipo       w1    w2    w3    w4    w5 
      ---------------------------------------     
      spam      3/6   5/6   4/6   6/6   6/6   
      legítimo  8/8   4/8   8/8   5/8   7/8      

     

#### Aplicación del Teorema de Bayes

A continuación se usará el teorema de Bayes para determine si el mensaje $w_1 w_4$ es spam. Es decir, se computará la probabilidad de que un mensaje conformado por las palabras $w_1 w_4$ sea spam.


Ya que este mensaje contiene las palabras $w_1$ y $w_4$ y no contiene las palabras $w_2$, $w_3$ y $w_5$, la probabilidad de que sea spam es:

$$\text{Pr}(\text{spam}~|~w_1~\text{and}~\text{not}~w_2~\text{and}~\text{not}~w_3~\text{and}~w_4~\text{and}~\text{not}~w_5)$$

La ecuación anterior no se puede calcular directamente a partir de las cuatro tablas de probababilidad anteriores (individuao, conjunta y condicionales), por lo que se debe recurrir al teorema de Bayes. De esta forma, se aplica  

$$\text{Pr}(B \; | \; A) = 
    \frac{\text{Pr}(A \; | \; B) \; \text{Pr}(B)}{\text{Pr}(A)}$$
    
    
de tal forma que la ecuación se transforma en:

$$\frac{\text{Pr}(w_1~\text{and}~\text{not}~w_2~\text{and}~\text{not}~w_3~\text{and}~w_4~\text{and}~\text{not}~w_5 |~\text{spam}) * \text{Pr}(\text{spam})}
{\text{Pr}(~w_1~\text{and}~\text{not}~w_2~\text{and}~\text{not}~w_3~\text{and}~w_4~\text{and}~\text{not}~w_5)}
$$


Si se tiene en cuenta que la ocurrencia de la palabras $w_1$, $w_2$, $w_3$, $w_4$ y $w_5$ son eventos independientes, es decir, que la ocurrencia de una palabra es independiente de la ocurrencia de las otras, entonces, el término $\text{Pr}(w_1~\text{and}~\text{not}~w_2~\text{and}~\text{not}~w_3~\text{and}~w_4 ~\text{and}~\text{not}~w_5|~\text{spam})$ puede aproximarse como:

$$
\text{Pr}(w_1~|~\text{spam})*
\text{Pr}(\text{not}~w_2~|~\text{spam})*
\text{Pr}(\text{not}~w_3|~\text{spam})*
\text{Pr}(w_4~|~\text{spam})*
\text{Pr}(\text{not}~w_5|~\text{spam})
$$

Estas cantidades ya fueron computadas en las tablas anteriores.

#### Cómputo numérico

El cálculo numérico es el siguiente. La probabilidad de:

$$\text{Pr}(\text{spam}~|~w_1~\text{and}~\text{not}~w_2~\text{and}~\text{not}~w_3~\text{and}~w_4~\text{and}~\text{not}~w_5)$$


es equivalente a:

$$\text{Pr}(B \; | \; A) = 
    \frac{\text{Pr}(A \; | \; B) \; \text{Pr}(B)}{\text{Pr}(A)}$$
    
donde: 

* $\text{Pr}(A \; | \; B) =
\text{Pr}(w_1~|~\text{spam})*
\text{Pr}(\text{not}~w_2~|~\text{spam})*
\text{Pr}(\text{not}~w_3|~\text{spam})*
\text{Pr}(w_4~|~\text{spam})*
\text{Pr}(\text{not}~w_5|~\text{spam}) =
(3/6) * (5/6) * (4/6) * (0/6) * (6/6) = 0
$

* $\text{Pr}(B) = \text{Pr}(\text{spam}) = 6/14$

* $\text{Pr}(A) = \text{Pr}(~w_1~\text{and}~\text{not}~w_2~\text{and}~\text{not}~w_3~\text{and}~w_4~\text{and}~\text{not}~w_5) = 
\text{Pr}(w_1)*
\text{Pr}(\text{not}~w_2)*
\text{Pr}(\text{not}~w_3)*
\text{Pr}(w_4)*
\text{Pr}(\text{not}~w_5) =
(3/14) * (9/14) * (12/14) * (3/14) * (13/14)
$

Entonces, 

$$\text{Pr}(\text{spam}~|~w_1~\text{and}~\text{not}~w_2~\text{and}~\text{not}~w_3~\text{and}~w_4~\text{and}~\text{not}~w_5) = 0$$


Al calcular la probabilidad de que el mensaje $w_1 w_4$ se legítimo, la probabilidad también da 0. Este problema se corrige usando el estimador de Laplace.

### Estimador de Laplace

Al construir la tabla de probabilidades de las ocurrencias de las palabras, es posible que una palabra $w_k$ aparezca únicamente en los mensajes válidos y no aparezca en los mensajes spam (o viceversa). De esta forma, si se calcula la probabilidad posterior de un nuevo mensaje que no la contiene, el resultado es cero para spam y uno para válido. Para prevernir esta situación, se hace que el conteo inicial no arranque en cero con el fin de que la probabilidad de ocurrencia sea siempre mayor que cero. Esto equivale a tener un mensaje para cada clase conformado por todas las palabras posibles.

Esto es:

     Suma =       5  7  4  5  3

#### Probabilidad de que el mensaje $w_1 w_4$ sea spam:

El cálculo numérico es el siguiente. La probabilidad de:

$$\text{Pr}(\text{spam}~|~w_1~\text{and}~\text{not}~w_2~\text{and}~\text{not}~w_3~\text{and}~w_4~\text{and}~\text{not}~w_5)$$

es equivalente a:

$$\text{Pr}(B \; | \; A) = 
    \frac{\text{Pr}(A \; | \; B) \; \text{Pr}(B)}{\text{Pr}(A)}$$
    
donde: 

* $\text{Pr}(A \; | \; B) =
\text{Pr}(w_1~|~\text{spam})*
\text{Pr}(\text{not}~w_2~|~\text{spam})*
\text{Pr}(\text{not}~w_3|~\text{spam})*
\text{Pr}(w_4~|~\text{spam})*
\text{Pr}(\text{not}~w_5|~\text{spam}) =
(4/11) * (9/11) * (8/11) * (1/11) * (10/11) = 13/727
$

* $\text{Pr}(B) = \text{Pr}(\text{spam}) = 11/24$

* $\text{Pr}(A) = \text{Pr}(~w_1~\text{and}~\text{not}~w_2~\text{and}~\text{not}~w_3~\text{and}~w_4~\text{and}~\text{not}~w_5) = 
\text{Pr}(w_1)*
\text{Pr}(\text{not}~w_2)*
\text{Pr}(\text{not}~w_3)*
\text{Pr}(w_4)*
\text{Pr}(\text{not}~w_5) =
(5/24) * (17/24) * (20/24) * (5/24) * (21/24) = 5/223
$

Entonces, 

$$\text{Pr}(\text{spam}~|~w_1~\text{and}~\text{not}~w_2~\text{and}~\text{not}~w_3~\text{and}~w_4~\text{and}~\text{not}~w_5) = (13/727) * (5/223) / (11/24) = 36.56%$$

(1/13) * (8/13) * (12/13) * (4/13) * (11/13) = 10/879
$

* $\text{Pr}(B) = \text{Pr}(\text{legitimo}) = 113/24$

* $\text{Pr}(A) = \text{Pr}(~w_1~\text{and}~\text{not}~w_2~\text{and}~\text{not}~w_3~\text{and}~w_4~\text{and}~\text{not}~w_5) = 
\text{Pr}(w_1)*
\text{Pr}(\text{not}~w_2)*
\text{Pr}(\text{not}~w_3)*
\text{Pr}(w_4)*
\text{Pr}(\text{not}~w_5) =
(5/24) * (17/24) * (20/24) * (5/24) * (21/24) = 5/223
$

Entonces, 

$$\text{Pr}(\text{spam}~|~w_1~\text{and}~\text{not}~w_2~\text{and}~\text{not}~w_3~\text{and}~w_4~\text{and}~\text{not}~w_5) = (10/879) * (5/223) / (13/24) = 27.49%$$


#### Decisión

Ya que la probabilidad de que el mensaje sea spam (36.56%) es mayor que la probabilidad de que sea legítimo (27.49%), entonces se clasifica el mensaje como spam.

### Implementación computacional en el lenguaje R

A continuación se presenta la solución del ejercicio anterior usando el lenguaje R.

```r
##
## Preparación
##
# install.packages("tm")
# install.packages("e1071")  # Naive Bayes

library(tm)
library(e1071)
```

```r
##
## Se preparan los datos. No se puede usar wi
## por que R descarta palabras de menos de
## dos letras
##        
texto <- c('ww1 ww3',  
           'ww1 ww2 ww1 ww3',  
           'ww2 ww4',  
           'ww4 ww5 ww2',  
           'ww2 ww4 ww2')

tipo  <- c('spam' ,    
           'spam',             
           'ham',      
           'ham',          
           'ham')

##
## Se crea un dataframe con los datos como factores
##
data <- data.frame(texto=texto, 
                   tipo=factor(tipo))


##
## Se crea un corpus que es una colección de documentos.
##
data_corpus <- VCorpus(VectorSource(data$text))


## 
## Para ver los documentos del corpus se usa lapply
##
lapply(data_corpus, as.character)
```

```r
##
## Se puede consultar un mensaje individual
##
inspect(data_corpus[[1]])
```

```r
##
## Se crea la matriz de términos del documento.
## Las filas corresponden a los mensajes..
## Las columnas corresponden a las palabras.
## El contenido representa la cantidad de veces que 
## aparece una palabra en un documento determinado.
##
data_dtm <- DocumentTermMatrix(data_corpus)

##
## A continuación se visualiza el contenido de la 
## matriz de términos del documento
##
inspect(data_dtm)
```

```r
##
## La matriz de términos puede convertirse en 
## una matriz de R
##
as.matrix(data_dtm)
```

```r
##
## Se convierte la frecuencia de ocurrencia a "Yes" y "No"
##
convert_counts <- 
function(x) {
    x <- ifelse(x > 0, "Yes", "No")
}
x <- apply(data_dtm, MARGIN = 2, convert_counts)

##
## Construye el clasificador
##
clf <- naiveBayes(x, data$tipo)

##
## Se imprime un resumen de parámetros
## del clasificador
##
clf
```

```r
##
## Se pronostica para los datos de entrenamiento
##
predict(clf, x)
```

```r
##
## Se crea un dataframe para un mensaje
## compuesto por las palabras w1 w4 
##
w1w4 <- data.frame(ww1 = c('Yes'),
                   ww2 = c('No'),
                   ww3 = c('No'),
                   ww4 = c('Yes'),
                   ww5 = c('No'))

##
## Se pronostica con el clasificador
##
predict(clf, w1w4)
```
