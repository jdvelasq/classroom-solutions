## Definiciones básicas

### Espacio muestral 

Es el conjunto $S$ de todos los posibles resultados de un experimento.

### Función de probabilidad

Dado un espacio muestral $S$ y un conjunto de eventos $A_i$ mutuamente exclusivos (los eventos no pueden ocurrir simultáneamente, por ejemplo, tener y no tener una enfermedad determinada al mismo tiempo) y colectivamente exhasutivos (eventos  cubren todo el universo posible; por ejemplo, los dos eventos posibles de padecer o no tener una enfermedad cubren todos los casos posibles), una [*función de probabilidad*](https://es.wikipedia.org/wiki/Función_de_probabilidad) es una función Pr cualquiera que:

* $\text{Pr}(A_i) \ge 0, \quad \forall A \in S$


* $\text{Pr}(S) = 1$.

### Variable aleatoria

Una [variable aletaoria](https://es.wikipedia.org/wiki/Variable_aleatoria) $X$ es una función que mapea el espacio muestral en los números reales.

### Función de distribución acumulada de $X$ o $F_X(x)$

Una [*función de distribución*](https://es.wikipedia.org/wiki/Función_de_distribución) acumulada se define como:

$$F_X(x) = \text{Pr}_X (X \le x), \quad \forall x$$

Tal que:

* $\lim_{x \to -\infty} F_X(x) = 0$ y $\lim_{x \to \infty} F_X(x) = 1$
* $F_X(x)$ es no una función no decreciente de x.
* $F_X(x)$ es continua por la derecha (existencia).

### Función de masa de probabilidad

La función de masa de probabilidad $f_X(s)$ de una variable aleatoria discreta $X$ es dada por:

$$f_X(x) = \text{Pr}(X=x), \qquad \forall x.$$

### Función de densidad de probabilidad

La [*función de densidad de probabilidad*](https://es.wikipedia.org/wiki/Función_de_densidad_de_probabilidad) $f_X(x)$ para una variable aleatoria continua $X$ es una función que satisface con:

$$F_X(x) = \int_{-\infty}^x f_X(t) dt, \quad \forall x.$$

### Valor esperado

El [*valor esperado*](https://es.wikipedia.org/wiki/Esperanza_matemática) o media de una variable aleatoria $g(X)$, denotado como $\text{E}g(X)$, es:

$$ \text{E}g(X) =
   \begin{cases}      
      \int_{-\infty}^\infty g(x)f_X(x), & \text{Si $X$ es continua}\\
      \\
      \sum_{x \in X} g(x)f_X(x) = g(x)\text{Pr}(X=x), & \text{Si $X$ es discreta}
   \end{cases}
$$

### Momentos de una distribución

Los [*momentos*](https://es.wikipedia.org/wiki/Momento_muestral) son valores numéricos que caracterizan una distribución de probabilidades. Si dos distribuciones tienen los mismos momentos son idénticas.  El $n$-ésimo momento se define como:

$$M_n = \text{E}(x^n) = \int u^n f_X(t) dt$$

* Media o valor esperado:

$$ \text{E}(x) = \mu_x=\int u·f_X(t) dt = M_1$$

* Varianza:

$$\text{V}(x)= \sigma^2_x = \text{E}[(x-\mu)^2]=M_2 - M_1^2$$


* Simetría:

$$\text{S}_x = \text{E}[(x-\mu)^3]=M_3 - 3M_2M_1 + 2M_1^3$$

* Curtosis:

$$\text{K}_x = \text{E}[(x-\mu)^4]=M_4 - 4M_3M_1 + 6M_2M_1^2 - 3M_1^4$$




**Interpretación gráfica**

<img src="images/funciones-teoricas-probabilidad/momentos-distribucion.jpg" width=400>

* La media o valor esperado señala el centro de masa de la distribución (su punto cental).

* La varianza indica la dispersión respecto a la media. Valores más altos indican distribuciones más gruesas (o anchas).

* La asimetría es un indicador de la simetría.

* La curtosis indica que tan anchas son las colas de los extremos de la distribución. Valores más altos indican colas más gruesas, y por tanto, mayor probabilidad de que se produzcan valores extremos.

## Principales disribuciones de probabilidad

### Distribución Uniforme Discreta

Una variable aletoria $X$ tiene una [distribución discreta uniforme](https://es.wikipedia.org/wiki/Distribución_uniforme_discreta) (1, $N$) si:

$$\text{Pr}(X=x|N) = \frac{1}{N}, \quad x=1,...,N$$

donde $N$ es un entero.

$$\text{E}(X) = \frac{N+1}{2}$$

$$\text{Var}(X) = \frac{(N+1)(N-1)}{12}$$

```r
##
## A continuación se la distribución de los valores
## de las caras de un dado.
##
x = seq(from = 1, to = 6, by = 1)
options(repr.plot.width = 6, repr.plot.height = 3)
plot(x,  dunif(x, min = 1, max = 6), type='h', pch = 22, col='red', ylab='')
```

### Distribución Normal

La función de densidad de probabilidad de una [distribución normal](https://es.wikipedia.org/wiki/Distribución_normal) con media $\mu$ y varianza $\sigma^2$ se define como:

$$f(x | \mu, \sigma^2) =\frac{1}{\sigma \sqrt{2\pi}} \exp \left\{ -\frac{1}{2} \frac{(x-\mu)^2}{\sigma^2} \right\}$$


Para la distribución normal:
* $M_1 = \mu$
* $M_2 = \sigma^2$
* $M_3 = 0$
* $M_4 = 3\sigma^2$

En la siguiente figura se observan diferentes distribuciones normales que varian en su media y varianza.

<img src="images/funciones-teoricas-probabilidad/dist-normal.jpg" width=700>

Todos los lenguajes de programación que tienen librerías estadísticas contienen funciones que permiten obtener diferentes valores relacionados con la función de densidad de probabilidad. Por ejemplo, en el lenguaje R:


* La función `dnorm` permite obtener el valor de la función de densidad de probabilidad para un punto $z$.


* La función `pnorm` permite obtener el valor del área bajo la curva (probabilidad acumulada) entre $-\infty$ y el valor de $z$ 


* La función `qnorm` permite obtener el valor de $z$ para una probabilidad acumulada conocida.

```r
##
## A continuación se grafican varias distribuciones
## normales que difieren en su media (mean) y su varianza (sd).
##
x = seq(from = -4, to = +4, by = 0.1)
options(repr.plot.width = 6, repr.plot.height = 4)
plot(x,  dnorm(x, mean= 0.0, sd=sqrt(0.2)), type='l', col='red', ylab='')
lines(x, dnorm(x, mean=-2.0, sd=sqrt(0.5)), type='l', col='blue')
lines(x, dnorm(x, mean= 0.0, sd=sqrt(1.0)), type='l', col='black')
```

**Propiedades útiles:**

Sean $x \sim \text{N}(\mu_x, \sigma^2_x)$ y $y \sim \text{N}(\mu_y, \sigma^2_y)$, entonces:

$$x + y \sim \text{N}\left(\mu_x + \mu_y, \sigma_x^2 + \sigma_y^2\right)$$

Sea $x \sim \text{N}(\mu_x, \sigma^2_x)$ y los números reales arbitrarios $a$ y $b$, entonces:

$$ax + b \sim \text{N}\left(a \mu_x + b, a^2 \sigma_x^2\right)$$

Sean $x \sim \text{N}(\mu_x, \sigma^2_x)$ y $\varepsilon \sim \text{N}(0, 1^2)$, entonces:

$$x  \sim \mu + \sigma \varepsilon$$

### Distribución Normal Estándar

La distribución normal estándar se obtiene al hacer $\mu=0$ y $\sigma=1$ en la distribución normal:

$$f(z) =\frac{1}{\sqrt{2\pi}} \exp \left\{ -\frac{1}{2} z^2 \right\}$$

**Intevalos de confianza.** Corresponden a los valores de $x$ para los cuales el área central de la distribución normal es igual al porcentaje especificado. En la imagen presentada a continuación se grafican los intervalos de confianza para $\pm \; 3 \sigma$, $\pm \; 2 \sigma$ y $\pm \; \sigma$, que equivalen a áreas del 99.73%, 95.44% y 68.26% respectivamente.

```r
## para +/- 2 * sigma
1.0 - 2 * pnorm(-2)
```

### Distribución $t$ de Student:

Sean $x_i,...,x_n \sim \text{N}(\mu, \sigma^2)$, entonces:

$$\frac{M_1 - \mu}{\sqrt{\frac{1}{n}M_2}} \sim t_{n-1}$$

Esta [distribución](https://es.wikipedia.org/wiki/Distribución_t_de_Student) tiene una forma similar a la distribución normal pero un poco más delgada y con colas más pesadas, haciéndose más similar a la distribución normal conforme $n$ aumenta. Históricamente fue derivada como la aproximación a la distribución de probabilidades de la media de $n$ datos provenientes de una distribución con media $\mu$ y varianza $\sigma^2$.

### Distribución Chi-cuadrada

Sea $z_i \sim \text{N}(0, 1)$, para $i=1,..., n$, entonces:

$$z_1^2 + z_2^2 + \ldots + z_n^2 \sim \chi_n^2$$

sigue una distribución [*chi-cuadrada*](https://es.wikipedia.org/wiki/Distribución_χ²) donde $n$ es la cantidad de grados de libertad.

```r
##
## A continuación se grafican distribuciones 
## para 3, 5 y 10 grados de libertad.
##
x = seq(from = 0, to = 20, by = 0.1)
options(repr.plot.width = 6, repr.plot.height = 4)
plot(x,  dchisq(x, df = 3), type='l', col='black', ylab='')
lines(x, dchisq(x, df = 5), type='l', col='blue')
lines(x, dchisq(x, df =10), type='l', col='red')
```

### Distribución F

Sean $x_i, y_j \sim \text{N}(0, 1)$, para $i=1,..., n$ y $j=1,...,m$, entonces la [*distribución F*](https://es.wikipedia.org/wiki/Distribución_F) se define como:

$$F = \frac{\sum_{i=1}^n x_i^2 / n}{\sum_{j=1}^m y_i^2} \sim F_{n,m}$$

donde $m$ y $n$ son los grados de libertad.

```r
##
## A continuación se grafican distribuciones 
## para n (df1) igual a 3, 5 y 10 y 
## m (df2) igual a 10.
##
x = seq(from = 0, to = 4, by = 0.02)
options(repr.plot.width = 6, repr.plot.height = 4)
plot(x,  df(x, df1=10, df2=10), type='l', col='black', ylab='')
lines(x, df(x, df1= 5, df2=10), type='l', col='blue')
lines(x, df(x, df1= 3, df2=10), type='l', col='red')
```

### Distribución de Bernoulli

En la [*distribución de Bernoulli*](https://es.wikipedia.org/wiki/Distribución_de_Bernoulli) se tiene un experiemento en el cual hay dos y solo dos resultados posibles. Una variable aleatoria $X$ sigue una distribución de Bernoulli de orden $p$ si:

$$ X =
   \begin{cases}
      1, & \text{con probabilidad $p$} \\
      0, &\text{con probabilidad $1-p$}
   \end{cases}
$$

con:

$$\text{E}(X) = 1p + 0(1-p)=p$$

$$\text{Var}(X) = (1-p)^2p + (0-p)^2(1-p)=p(1-p)$$

### Distribución Binomial

La [*distribución binomial*](https://es.wikipedia.org/wiki/Distribución_binomial) se obtiene al considerar una secuencia de $n$ experimentos independientes de Bernoulli (véase la distribución de Bernoulli), es decir, se toman $n$ varaibles aleatorias $X$ que pueden tomar el valor de 1 (con probabilidad $p$) o de 0 (con probabilidad $1-p$),  y se obtienen exactamente $Y$ éxitos, es decir, ocurre $X=1$, $y$ veces. En otras palabras, si se tiene una secuencia de $n$ experimentos idénticos, cada uno con una probabilidad de éxito $p$, y se definen las variables aleatorias $X_1$, ..., $X_n$ como:

$$ X_i =
   \begin{cases}
      1, & \text{con probabilidad $p$} \\
      0, &\text{con probabilidad $1-p$}
   \end{cases}
$$

La variable aleatoria

$$Y = \sum_{i=1}^n X_i$$

sigue una distribución binomial($n$, $p$).



```r
##
## A continuación se grafica la distribición
## para 20 experimentos (n), con una probabilidad
## p igual a 0.5
##
x = seq(from = 0, to = 20, by = 1)
options(repr.plot.width = 6, repr.plot.height = 4)
plot(x,  dbinom(x, size=20, prob=0.5), type='l', col='black', ylab='')
```

### Distribución Poisson

Esta distribución se usa para modelar el número de ocurrencias de un evento en un intervalo de tiempo. Una variable $X$ sigue una distribución de Poisson($\lambda$), si:

$$P(X=x|\lambda) = \frac{\exp(-\lambda) * \lambda^x}{x!}, \quad x=0,1,...$$

con:

$$\text{E}(X)=\text{Var}(X) = \lambda$$

### Distribución Exponencial

En la [*distribución exponencial*](https://es.wikipedia.org/wiki/Distribución_exponencial):

$$f(X|\beta)=\frac{1}{\beta}\exp \left( \frac{-x}{\beta} \right)$$

La cual es usada para modelar tiempos de vida.

$$\text{E}(X)= \beta$$

$$\text{Var}(X) = \beta^2$$

```r
##
## A continuación se grafica la distribición
## la distribución para lambda = 1.5, 1.0 y 0.5
##
x = seq(from = 0, to = 5, by = 0.1)
options(repr.plot.width = 6, repr.plot.height = 4)
plot(x,  dexp(x, rate=1.5), type='l', col='black', ylab='')
lines(x, dexp(x, rate=1.0), type='l', col='blue')
lines(x, dexp(x, rate=0.5), type='l', col='red')
```

### Distribución de Weibull

La distribución de [*Weibull*](https://es.wikipedia.org/wiki/Distribución_de_Weibull) se usa para modelar la distribución de fallos en sistemas cuando la tasa de fallos es proporcional a una potencia del tiempo. Se define como:

$$f(X|\lambda, k) = 
   \begin{cases}
      \frac{k}{\lambda} \left( \frac{X}{\lambda} \right)^{k-1} 
      \exp \left( - \left(\frac{x}{\lambda} \right)^k \right), & x \ge 0 \\
      0, & x < 0
   \end{cases}
$$

* $k < 1$ indica que la tasa de fallos decrece con el tiempo.
* $k=1$ la tasa de fallos es constante en el tiempo.
* $k>1$ indica que la tasa de fallos crece en el tiempo.
