===

---

## Definición del problema real

EPS SURA desea mitigar los riesgos asociados a enfermedades que generan altos costos para la compañía, representados en costos de medicamentos, hospitalización, cirugias y cuidados. De acuerdo con estudios médicos, se presume que algunas de estas enfermedades guardan correlación con medidas de algunas sustancias en el cuerpo como el nivel de colesterol, nivel de hemoglobina, nivel de trigliceridos, entre otros, y también con la presencia de otras enfermedades. En este caso, se desea predecir si un paciente tendrá o no ERC (Enfermedad Renal Crónica) de acuerdo a sus características.

## Definición del problema en términos de los datos

Se tiene una muestra de aproximadamente 88.200 observaciones. Cada registro contiene 24 atributos que caracterizan a un paciente en términos de historial médico y niveles de algunas sustancias en su cuerpo, aclarando si presenta o no ERC. La información se obtuvo de la base de datos de pacientes de la EPS. 

De acuerdo con el problema planteado, se define **Ind_ERC** como la variable objetivo. A continuación se realiza una exploración de los datos con el objetivo de definir que variables son convenientes para establecer el diagnóstico de un paciente.

## Exploración

```r
summary(y)
```

Note que los niveles de proteinuria y ácido úrico presentan mas de 70.000 datos faltantes, lo cual en términos del modelo, podría disminuir la muestra significativamente, casi en un 80%. Es necesario observar si apriori estos atributos guardan correlación con la variable de decisión.

```r
##
## Contiene 72.344 datos faltantes
##
summary(y$Ind_Proteinuria)
```

```r
##
## Contiene 76.305 datos faltantes
##
summary(y$Ind_Acido_Urico)
```

A continuación se trata de ajustar una regresión simple entre la variable de decisión y los niveles de sustancia mencionados:

```r
##
## Regresión entre Ind_ERC e Ind_Proteinuria
## El valor p indica que es significativo
##
pre_model <- lm(Ind_ERC ~ Ind_Proteinuria, data = y)
summary(pre_model)
```

```r
##
## Regresión entre Ind_ERC e Ind_Acido_Urico
## El valor p indica que es significativo
##
pre_model <- lm(Ind_ERC ~ Ind_Acido_Urico, data = y)
summary(pre_model)
```

De acuerdo con los anteriores resultados es posible concluir que no existe evidencia suficiente para extraer los atributos *Ind_Acido_Urico* e *Ind_Proteinuria* del modelo, por lo que se opta por construir un modelo que contenga dichas variables con una cantidad menor de transacciones y uno análogo que no las contemple pero que contenga la cantidad máxima de registros con la información completa.

```r
str(y)
```

```r
## 
## Las datos faltantes no se encuentran para los mismos pacientes. Por ello se disminuye el número de registros en tal magnitud. 
##
data <- y[,c(-14:-16,-20)]
data_aux <- y[,c(-14,-15)]

data <-  data[complete.cases(data),]
data_aux <-  data_aux[complete.cases(data_aux),]

str(data)
str(data_aux)
```

```r
##
## Se cambian las variables binarias a factores.
##
data[,c(colnames(data)[3:13])] <- lapply(data[,c(colnames(data)[3:13])] , factor)
data[,c('Ind_ERC')] <- as.factor(data[,c('Ind_ERC')])


data_aux[,c(colnames(data_aux)[3:13])] <- lapply(data_aux[,c(colnames(data_aux)[3:13])] , factor)
data_aux[,c('Ind_ERC')] <- as.factor(data_aux[,c('Ind_ERC')])
```

## Metodología 
Para resolver el modelo se tendrán en cuenta diferentes clasificadores, iniciando el algoritmo OneR que brindará la métrica base para evaluar los modelos, luego se tendrá un cuenta un árbol de clasifiación simple seguido de adaptive boosting y por último un modelo de bosque aleatorio. Se escogerá el modelo que reporte la mejor precisión a nivel de pronóstico.

## Modelo 1. Sin tener en cuenta ind_acido_urico e ind_proteinuria

### One R

```r
library(OneR)
clf <- OneR(Ind_ERC ~ ., data = data, verbose = TRUE)
clf
```

Note que en el modelo OneR, todas las reglas predicen que lo pacientes son sanos y logra una precisión inicial del 83,95%. Se toma esta medida como métrica de predicción.

## Preparación de los datos

```r
##
## Se cargan las librerías necesarias
##
library(caret)
library(lattice)
library(ggplot2)
```

Para los siguientes modelos, se tendrá en cuenta un set de entrenamiento del 75% y un set de prueba que evaluará el modelo en términos de pronóstico del 25%. Se utilizará un muestreo estratificado debido a que los datos se encuentran desbalanceados con respecto al número de casos que presentan ERC, es decir, se debe tomar, en proporción, la misma cantidad de datos de pacientes que padecen la enfermedad y que no.  

```r
set.seed(300)
train.index <- createDataPartition(data$Ind_ERC, p = .75, list = FALSE)

#Datos de entrenamiento 
x_train <- data[train.index,-19]                    # Dataframe con los datos de entrenamiento
y_train_labels <- as.factor(data[train.index,19])   # Vector que contiene los valores de la variable de decisión para el entrenamiento

#Datos de orueba
x_test <- data[-train.index,-19]
y_test_labels <- as.factor(data[-train.index,19])

#Datos de entrenamiento y prueba completos 
x_train_completed <- data[train.index,]
x_test_completed <- data[-train.index,]
```

```r
##
## Proporción de los datos para entrenamiento y prueba
##
prop.table(table(y_train_labels))
prop.table(table(y_test_labels))
```

### Árbol de clasificación

**Entrenamiento del modelo**

Se utilizará la librería C50 para implementar un árbol simple de clasificación.

```r
modelLookup('C5.0')
```

```r
library(C50)
clf <- C5.0(x_train, y_train_labels, winnow=TRUE)
clf
```

```r
summary(clf)
```

El error de entrenamiento es aproximadamente 3%.

**Evaluación del modelo**

```r
library(gmodels)
y_test_pred <- predict(clf, x_test)

CrossTable(y_test_labels, 
           y_test_pred,
           prop.chisq = FALSE, 
           prop.c = TRUE, 
           prop.r = FALSE,
           dnn = c('Real', 'Pronóstico'))
```

```r
##
## Error predicción
##
error <- round(((1-((5905+901)/7034))*100),4)
paste(error,'%')
```

### Adaptive Boosting

**Entrenamiento del modelo**

Se entrena el modelo con 10 árboles.

```r
clf10 <- C5.0(x_train, 
              y_train_labels,
              trials = 10)  
clf10
```

```r
summary(clf10)
```

El error de entrenamiento es del 3.9% aproximadamente.

**Evaluación del modelo**

```r
library(gmodels)
y_test_pred <- predict(clf10, x_test)

CrossTable(y_test_labels, 
           y_test_pred,
           prop.chisq = FALSE, 
           prop.c = TRUE, 
           prop.r = FALSE,
           dnn = c('Real', 'Pronóstico'))
```

```r
##
## Error predicción
##
error <- round(((1-((5785+959)/7034))*100),4)
paste(error,'%')
```

### Random Forest

```r
library('caret')
modelLookup('rf')
```

**Entrenamiento del modelo**

Se prueban los posibles predictores entre todas las variables del modelo.

```r
library(randomForest)

traincontrol <- trainControl(method = "repeatedcv", number = 10, repeats = 3, search = "grid")

set.seed(300)
tunegrid <- expand.grid(.mtry=c(1:18))
rf_tree <- train(Ind_ERC ~ ., data = x_train_completed, method = 'rf', tuneGrid=tunegrid)
rf_tree
```

**Evaluación del modelo**

```r
rf_tree_pred <- predict(rf_tree, x_test)
confusionMatrix(rf_tree_pred, y_test_labels)
```
